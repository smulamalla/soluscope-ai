```python
from rdkit import Chem
from rdkit.Chem import Descriptors, Lipinski

smiles = "CC(=O)OC1=CC=CC=C1C(=O)O" # aspirin
mol = Chem.MolFromSmiles(smiles)

properties = {
    "Molecular Weight": Descriptors.MolWt(mol),
    "LogP": Descriptors.MolLogP(mol),
    "H-Bond Donors": Lipinski.NumHDonors(mol),
    "H-Bond Acceptors": Lipinski.NumHAcceptors(mol),
    "Rotatable Bonds": Descriptors.NumRotatableBonds(mol),
    "TPSA": Descriptors.TPSA(mol)
}

for name, value in properties.items():
    print(f"{name}: {value:.2f}")
```

```python
def check_lipinski(mol):
    violations = 0
    if Descriptors.MolWt(mol) > 500: violatons += 1
    if Descriptors.MolLogP(mol) > 5: violations += 1
    if Lipinski.NumHDonors(mol) > 5: violations += 1
    if Lipinski.NumHAcceptors(mol) > 10: violations += 1
    return violations

violations = check_lipinski(mol)
print(f"Lipinski violations: {violations} ({'Drug-like' if violations <=1 else 'Possibly not drug-like'})")
```

```python
def get_molecular_properties(smiles: str) -> dict:
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        raise ValueError(f"Invalid SMILES: {smiles}")
    return {
        "Molecular Weight": round(Descriptors.MolWt(mol), 2),
        "LogP": round(Descriptors.MolLogP(mol), 2),
        "H-Bond Donors": Lipinski.NumHDonors(mol),
        "H-Bond Acceptors": Lipinski.NumHAcceptors(mol),
        "Rotatable Bonds": Descriptors.NumRotatableBonds(mol),
        "TPSA": round(Descriptors.TPSA(mol), 2),
        "Lipinksi Violations": check_lipinski(mol)
    }

print(get_molecular_properties("CCO"))
```
