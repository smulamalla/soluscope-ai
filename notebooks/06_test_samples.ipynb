```python
import deepchem as dc
import numpy as np

# Load dataset
tasks, datasets, transformers = dc.molnet.load_delaney(featurizer="ECFP", splitter="random")

train_dataset, valid_dataset, test_dataset = datasets

# Recreate model architecture
model = dc.models.MultitaskRegressor(
    n_tasks=1,
    n_features=1024,
    layer_sizes=[500, 100],
    model_dir="../models/multitask_regressor"
)

# Load trained weights
model.restore()
```

```python
sample_smiles =  [
    "CCO", # ethanol - very soluble
    "c1ccccc1", # benzene - poorly soluble
    "CC(=O)OC1=CC=CC=C1C(=O)O", # aspirin
    "C1=CC=C2C(=C1)C=CC3=CC=CC=C23" # anthracene - very insoluble
]
```

```python
from deepchem.feat import CircularFingerprint

featurizer = CircularFingerprint(size=1024, radius=2)
features = featurizer.featurize(sample_smiles)

sample_dataset = dc.data.NumpyDataset(X=features)
predictions = model.predict(sample_dataset)
```

```python
untransformed_preds = transformers[0].untransform(predictions).flatten()

for smiles, pred in zip(sample_smiles, untransformed_preds):
    print(f"{smiles:40s} -> predicted log solubility: {pred:.2f}")
```
