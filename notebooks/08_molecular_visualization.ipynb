```python
from rdkit import Chem
from rdkit.Chem import Draw

smiles = "CC(=O)OC1=CC=CC=C1C(=O)O" # aspirin
mol = Chem.MolFromSmiles(smiles)

img = Draw.MolToImage(mol, size=(300, 300))
img
```

```python
sample_smiles = ["CCO", "c1ccccc1", "CC(=O)OC1=CC=CC=C1C(=O)O", "C1=CC=C2C(=C1)C=CC3=CC=CC=C23"]
mols = [Chem.MolFromSmiles(s) for s in sample_smiles]
labels = ["Ethanol", "Benzene", "Aspirin", "Anthracene"]

grid_img = Draw.MolsToGridImage(mols, molsPerRow=4, subImgSize=(200, 200), legends=labels)
grid_img
```
