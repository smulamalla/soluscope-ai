```python
import deepchem as dc
import pickle
import numpy as np
from sklearn.ensemble import RandomForestRegressor

with open("../models/dataset_split.pkl", "rb") as f:
    train_dataset, valid_dataset, test_dataset, transformers = pickle.load(f)
```

```python
rf = RandomForestRegressor(n_estimators=100)
rf.fit(train_dataset.X, train_dataset.y.ravel())

with open("../models/random_forest.pkl", "wb") as f:
    pickle.dump(rf, f)
```

```python
top_bits = np.argsort(rf.feature_importances_)[::-1][:10]

with open("../models/top_bits.pkl", "wb") as f:
    pickle.dump(top_bits, f)
```
