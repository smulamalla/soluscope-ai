```python
import deepchem as dc
import pickle

tasks, datasets, transformers = dc.molnet.load_delaney(featurizer='ECFP', splitter=None)
dataset = datasets[0]

# Create one reproducible random split
splitter = dc.splits.RandomSplitter()

train_dataset, valid_dataset, test_dataset = splitter.train_valid_test_split(
    dataset,
    frac_train=0.8,
    frac_valid=0.1,
    frac_test=0.1,
    seed=42
)

# Save the exact split for later notebooks
with open("../models/dataset_split.pkl", "wb") as f:
    pickle.dump(
        (train_dataset, valid_dataset, test_dataset, transformers),
        f
    )
print("Train:", len(train_dataset))
print("Valid:", len(valid_dataset))
print("Test:", len(test_dataset))
```

```python
model = dc.models.MultitaskRegressor(
    n_tasks=len(tasks),
    n_features=1024,
    layer_sizes=[500, 100],
    model_dir="../models/multitask_regressor"
)
```

```python
model.fit(train_dataset, nb_epoch=100)
```

```python
metric = dc.metrics.Metric(dc.metrics.pearson_r2_score)
train_score = model.evaluate(train_dataset, [metric], transformers)
print("Train R²:", train_score)
```
