```python
import deepchem as dc

tasks, datasets, transformers = dc.molnet.load_delaney(featurizer='Raw', splitter='random')
train_dataset, valid_dataset, test_dataset = datasets

print(tasks)
print(train_dataset)
```

```python
print("Train size:", len(train_dataset))
print("Valid size:", len(valid_dataset))
print("Test size:", len(test_dataset))

mol = train_dataset.X[0]
label = train_dataset.y[0]
print("Molecule:", mol)
print("Solubility (log mol/L):", label)
```

```python
total = len(train_dataset) + len(valid_dataset) + len(test_dataset)
print(f"Train: {len(train_dataset)/total:.0%}, Valid: {len(valid_dataset)/total:.0%}, Test: {len(test_dataset)/total:.0%}")
```
